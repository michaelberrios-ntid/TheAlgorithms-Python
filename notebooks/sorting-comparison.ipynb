## System Under Test:
This notebook will test the system by mimicking a potential user of TheAlgorithms/Python

### Context:
A user is investigating different sorting algorithms and wants to run tests to compare their speeds.
They create a file and import the relevant libraries and methods.

```python
import sys
print(sys.executable)
```

```python

import random
import time

from sorts.bubble_sort import bubble_sort_iterative
from sorts.bucket_sort import bucket_sort
from sorts.circle_sort import circle_sort
from sorts.cocktail_shaker_sort import cocktail_shaker_sort
from sorts.heap_sort import heap_sort
from sorts.insertion_sort import insertion_sort
from sorts.merge_sort import merge_sort
from sorts.natural_sort import natural_sort
from sorts.odd_even_sort import odd_even_sort
from sorts.pancake_sort import pancake_sort
from sorts.pigeonhole_sort import pigeonhole_sort
from sorts.selection_sort import selection_sort
from sorts.quick_sort import quick_sort
from sorts.stalin_sort import stalin_sort
from sorts.tim_sort import tim_sort
from sorts.tree_sort import tree_sort
from sorts.binary_insertion_sort import binary_insertion_sort
from sorts.comb_sort import comb_sort
from sorts.gnome_sort import gnome_sort
from sorts.iterative_merge_sort import iter_merge_sort
from sorts.merge_insertion_sort import merge_insertion_sort
from sorts.stooge_sort import stooge_sort
from sorts.wiggle_sort import wiggle_sort

from tabulate import tabulate
```

The user then creates datasets of various sizes to test on, as well as lists to hold results.


```python
random.seed(777)
hundred = random.sample(range(1, 101), 100)
thousand = random.sample(range(1, 1001), 1000)
tenThousand = random.sample(range(1, 10001), 10000)
num_lists = [hundred, thousand, tenThousand]

sorting_methods = [bubble_sort_iterative, bucket_sort, circle_sort,cocktail_shaker_sort,
                   binary_insertion_sort, comb_sort, gnome_sort,heap_sort, insertion_sort,
                   iter_merge_sort, merge_insertion_sort, merge_sort,natural_sort, odd_even_sort,
                   pancake_sort, pigeonhole_sort, quick_sort, selection_sort, tim_sort, tree_sort,
                   stooge_sort, wiggle_sort, stalin_sort]
results = {method : [] for method in sorting_methods}
```

The user then runs the lists of numbers through each method and records the times of completion.
They also check that all sorting methods work properly.

```python
for method in sorting_methods:
    for nums in num_lists:
        test_case = nums.copy()
        start = time.perf_counter()
        method(test_case)
        end = time.perf_counter()
        results[method].append(end - start)
        print(f'{method.__name__}, Length: {len(test_case)}, Time: {end - start:.4f}')
        if test_case != sorted(nums):
            print(f'{method.__name__} failed!' )
```

After gathering time data, the user organizes it into a table to derive conclusions.

```python
tabled_results = [[method.__name__, results.get(method, [])[0],
                   results.get(method, [])[1],
                   results.get(method, [])[2]] for method in sorting_methods]
headers = ["Method", "100 numbers", "1000 numbers", "10000 numbers"]

print(tabulate(tabled_results, headers, tablefmt="grid"))
```

## Outcome
User successfully tested various sorting methods and can compare them based on their execution times.
